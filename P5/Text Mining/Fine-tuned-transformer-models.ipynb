# How to use fine-tuned transformer models

available tasks are ['audio-classification', 'automatic-speech-recognition', 'conversational', 'feature-extraction', 'fill-mask', 'image-classification', 'image-segmentation', 'ner', 'object-detection', 'question-answering', 'sentiment-analysis', 'summarization', 'table-question-answering', 'text-classification', 'text-generation', 'text2text-generation', 'token-classification', 'translation', 'zero-shot-classification', 'translation_XX_to_YY']"

```python
from transformers import pipeline
```

```python
nerc_classifier = pipeline("ner")
```

```python
result = nerc_classifier("Hello I'm Ilia, I work at the VU University and I don't live in Weesp.")
```

```python
print(result)
```

```python
sentiment_classifier = pipeline("sentiment-analysis")
```

```python
result = sentiment_classifier("I really like Weesp. It is close to Amsterdam, has an old center and close to nature.")
```

```python
print(result)
```

```python
nerc_classifier = pipeline("ner", model = "xlm-roberta-large-finetuned-conll03-english")
```

```python
result = nerc_classifier("Hallo ik ben Ilia, ik werk aan de Vrije Universiteit en ik woon niet in Weesp.")
```

```python
print(result)
```

```python
result = nerc_classifier("Apples sued Samsung for patents last year.")
```

```python
print(result)
```
